```python
import itertools

def evaluate_sentence(sentence, model):
    """
    Evaluates a logical sentence under a given model (truth assignment).
    Supports variables, negation (~), conjunction (&), disjunction (|), implication (>>), and equivalence (==).
    """
    try:
        return eval(sentence, {}, model)
    except Exception as e:
        print(f"Error evaluating sentence '{sentence}': {e}")
        return False

def tt_entails(KB, alpha, symbols):
    """
    Truth-Table Enumeration algorithm to decide if KB entails alpha.
    Returns True if KB entails alpha, False otherwise, along with the full truth table.
    """
    combinations = list(itertools.product([True, False], repeat=len(symbols)))
    truth_table_results = []
    entails = True

    for combo in combinations:
        model = dict(zip(symbols, combo))
        kb_val = True
        for clause in KB:
            if not evaluate_sentence(clause, model):
                kb_val = False
                break

        query_val = evaluate_sentence(alpha, model)

        if kb_val and not query_val:
            entails = False

        truth_table_results.append((model, kb_val, query_val))

    return entails, truth_table_results

def display_truth_table(symbols, KB, alpha, results):
    """
    Helper function to print a clean truth table.
    """
    header = list(symbols) + ["KB", "Query", "KB => Query (Valid?)"]
    print(" | ".join(f"{col:<8}" for col in header))
    print("-" * (11 * len(header)))

    for model, kb_val, query_val in results:
        row_vals = [str(model[sym]) for sym in symbols]
        valid = "YES" if (not kb_val or query_val) else "NO (FAIL)"
        row_vals += [str(kb_val), str(query_val), valid]
        print(" | ".join(f"{val:<8}" for val in row_vals))

if __name__ == "__main__":
    symbols = ['P', 'Q']

    KB = [
        '(not P) or Q',  # P => Q
        'P'              # P
    ]
    query = 'Q'

    entails, results = tt_entails(KB, query, symbols)

    print(f"Does KB entail the query '{query}'? : {entails}\n")
    display_truth_table(symbols, KB, query, results)
```
