```python
INITIAL_STATE = (5, 4, 6, 1, 8, 7, 3, 2, 0)
GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def print_board(state):
    """Utility function to print the 3x3 puzzle board layout."""
    for i in range(0, 9, 3):
        print(f"{state[i]} {state[i+1]} {state[i+2]}")
    print("-" * 10)

def get_neighbors(state):
    """Generates valid next states by shifting tiles into the blank space (0)."""
    neighbors = []
    blank_idx = state.index(0)
    row, col = divmod(blank_idx, 3)

    moves = [(-1, 0, -3), (1, 0, 3), (0, -1, -1), (0, 1, 1)]

    for dr, dc, delta in moves:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            neighbor = list(state)
            neighbor[blank_idx], neighbor[blank_idx + delta] = neighbor[blank_idx + delta], neighbor[blank_idx]
            neighbors.append(tuple(neighbor))

    return neighbors

def depth_limited_search(state, goal, limit, visited):
    """
    Performs Depth-Limited Search (DLS).
    Returns a list representing the path if successful, 'cutoff' if limit reached, or None.
    """
    if state == goal:
        return [state]
    if limit == 0:
        return "cutoff"

    cutoff_occurred = False
    visited.add(state)

    for neighbor in get_neighbors(state):
        if neighbor not in visited:
            result = depth_limited_search(neighbor, goal, limit - 1, visited)

            if result == "cutoff":
                cutoff_occurred = True
            elif result is not None:
                return [state] + result

    visited.remove(state)
    return "cutoff" if cutoff_occurred else None

def iterative_deepening_search(initial, goal, max_depth=50):
    """Repeatedly invokes DLS with expanding depth limits (IDS)."""
    for depth in range(max_depth):
        visited = set()
        result = depth_limited_search(initial, goal, depth, visited)

        if result != "cutoff" and result is not None:
            print(f"Goal reached at depth limit: {depth}")
            return result
    return None

print("Initial Board State:")
print_board(INITIAL_STATE)

print("Searching for solution via Iterative Deepening (DLS updates)...")
solution_path = iterative_deepening_search(INITIAL_STATE, GOAL_STATE)

if solution_path:
    print(f"\nSolution Found! Total steps taken: {len(solution_path) - 1}\n")
    for step, state in enumerate(solution_path):
        print(f"Step {step}:")
        print_board(state)
else:
    print("No solution found within the maximum depth constraint.")
```
